# Module B2 - NLP Sentiment Analysis & Vectorization Training
This notebook demonstrates text preprocessing (lowercasing, stopword removal, tokenization) and TF-IDF + Logistic Regression training.

```python
import pandas as pd
import numpy as np
import joblib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

df = pd.read_csv('../data/reviews.csv')
print(f'Loaded {len(df)} customer reviews.')
df.head()
```

## 1. Vectorization & Sentiment Classifier Training

```python
vectorizer = TfidfVectorizer(ngram_range=(1, 2), min_df=1)
X_tfidf = vectorizer.fit_transform(df['review_text'])
y_labels = df['sentiment']

clf = LogisticRegression(C=1.0, max_iter=200)
clf.fit(X_tfidf, y_labels)

y_preds = clf.predict(X_tfidf)
print(classification_report(y_labels, y_preds))
```

## 2. Export Artifacts

```python
joblib.dump(clf, '../app/models/sentiment_model.pkl')
joblib.dump(vectorizer, '../app/models/vectorizer.pkl')
print('Exported sentiment_model.pkl and vectorizer.pkl successfully!')
```
