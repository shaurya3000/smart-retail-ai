# Module A2 - Product Image Category Classifier Training
This notebook demonstrates the end-to-end training pipeline for the retail product category classifier.

```python
import numpy as np
import pandas as pd
import joblib
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

print('Libraries imported successfully!')
```

## 1. Feature Representation & Dataset Simulation
We train on 5 retail product categories: Clothing, Shoes, Electronics, Bags, Groceries.

```python
categories = ['clothing', 'shoes', 'electronics', 'bags', 'groceries']
np.random.seed(42)
X_data = []
y_data = []

for idx, cat in enumerate(categories):
    mean_vec = np.zeros(32)
    mean_vec[idx * 5 : (idx + 1) * 5] = 2.5 + idx * 0.8
    mean_vec[25 + idx] = 1.8
    samples = np.random.normal(loc=mean_vec, scale=0.4, size=(200, 32))
    X_data.append(samples)
    y_data.extend([cat] * 200)

X_data = np.vstack(X_data)
y_data = np.array(y_data)
print(f'Generated dataset with shape {X_data.shape}')
```

## 2. Model Training & Evaluation

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_data, y_data, test_size=0.2, random_state=42)

clf = RandomForestClassifier(n_estimators=100, random_state=42)
clf.fit(X_train, y_train)

y_pred = clf.predict(X_test)
print(classification_report(y_test, y_pred))
```

## 3. Serialization to Model File

```python
joblib.dump(clf, '../app/models/product_classifier.pkl')
print('Product classifier exported successfully!')
```
