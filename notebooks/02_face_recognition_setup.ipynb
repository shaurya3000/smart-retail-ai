# Module A3 - Face Recognition & Customer Visit Logging Setup
This notebook demonstrates face detection, 128D encoding extraction, matching algorithms, and visit logging.

```python
import numpy as np
import pickle
import cv2

print('OpenCV & NumPy initialized!')
```

## 1. Customer Database Registration
Generating facial feature vectors for consenting customer profiles.

```python
customers = [
    {'customer_id': 'CUST_1001', 'name': 'Alice Smith', 'loyalty_tier': 'Gold'},
    {'customer_id': 'CUST_1002', 'name': 'Bob Johnson', 'loyalty_tier': 'Platinum'}
]

face_db = []
for idx, cust in enumerate(customers):
    np.random.seed(100 + idx)
    enc = np.random.normal(loc=0.0, scale=1.0, size=128)
    enc = enc / np.linalg.norm(enc)
    face_db.append({
        'customer_id': cust['customer_id'],
        'name': cust['name'],
        'loyalty_tier': cust['loyalty_tier'],
        'encoding': enc.tolist()
    })

print(f'Registered {len(face_db)} customer profiles.')
```

## 2. Cosine Distance Matching Algorithm

```python
def match_face(query_vec, db):
    best_match = None
    min_dist = 999.0
    for record in db:
        stored_enc = np.array(record['encoding'])
        dist = 1.0 - (np.dot(query_vec, stored_enc) / (np.linalg.norm(query_vec) * np.linalg.norm(stored_enc)))
        if dist < min_dist:
            min_dist = dist
            best_match = record
    return best_match, min_dist

test_query = np.array(face_db[0]['encoding']) + np.random.normal(0, 0.05, 128)
match, dist = match_face(test_query, face_db)
print(f'Matched customer: {match["name"]}, Cosine Distance: {dist:.4f}')
```
